```python
"""
Regularization:到底在限制什么
    不要让模型只寻找最容易把训练集做对的解,而是让它更倾向于学习能够推广到新的数据上
"""
```

```python
"""
网络变深后,数值尺度也会不断传播
"""
```

```python
"""
Normalization在稳定什么
    先选择一组元素计算统计量,再利用这些统计量重新调整特征的尺度
"""
```

```python
"""
Normalization 不是简单的防止梯度消失
    Normalization 通过控制中间表示的统计尺度,使深层网络中数值传播和参数优化更加可控,不同Normalization的具体效果和机制也不相同
"""
```
